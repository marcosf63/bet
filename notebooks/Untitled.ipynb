```python
import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings("ignore")

def drop_reset_index(df):
    df = df.dropna()
    df = df.reset_index(drop=True)
    df.index += 1
    return df
```

```python
dia = "2025-08-17"
url = f"https://github.com/futpythontrader/Jogos_do_Dia/raw/refs/heads/main/FootyStats/Jogos_do_Dia_FootyStats_{dia}.csv"
url2= f'https://github.com/futpythontrader/Jogos_do_Dia/raw/refs/heads/main/FlashScore/Jogos_do_Dia_FlashScore_{dia}.csv'
```

```python
Jogos_do_Dia_FootyStats = pd.read_csv(url2)
Jogos_do_Dia_FootyStats = drop_reset_index(Jogos_do_Dia_FootyStats)
```
