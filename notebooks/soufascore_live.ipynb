```python
import requests
from pydantic import BaseModel
from typing import Dict
```

```python
resp = requests.get('https://www.sofascore.com/api/v1/sport/football/events/live')
```

```python
lives = resp.json()
```

```python
events = lives['events']
```

```python
events[20]['tournament']['category']['country']['name']
```

```python
class Tournament(BaseModel):
    name: str
    category: Dict

class Status(BaseModel):
    description: str
    type: str

class HomeTeam(BaseModel):
    name: str
    slug: str
    shortName: str
    
class AwayTeam(BaseModel):
    name: str
    slug: str
    shortName: str

class HomeScore(BaseModel):
    current: int

class AwayScore(BaseModel):
    current: int

class Event(BaseModel):
    tournament: Tournament
    status: Status
    homeTeam: HomeTeam
    awayTeam: AwayTeam
    homeScore: HomeScore
    awayScore: AwayScore
    
    
```

```python
event_objects = [Event(**event) for event in events]
```

```python
event_objects[0]
```

```python
for jogo in event_objects:
    if jogo.homeScore.current  + jogo.awayScore.current == 1:
        print(f'{jogo.homeTeam.name} {jogo.homeScore.current} x {jogo.awayScore.current} {jogo.awayTeam.name}')
```
