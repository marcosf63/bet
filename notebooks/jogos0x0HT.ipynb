```python
df = pd.read_csv("./dados/Base_de_Dados_Betfair Exchange_Filtrada_2025-07-02.csv")
```

```python
df = df[["Date", "League", "Home", "Away", "Goals_H_FT", "Goals_A_FT", "Goals_Min_H","Goals_Min_A", "Odd_H_FT", "Odd_D_FT", "Odd_A_FT", "Odd_Over25_FT", "Odd_BTTS_Yes"]]
```

```python
import pandas as pd
import numpy as np

# Função que garante lista de minutos (mesmo que vazia)
def garantir_lista(val):
    if isinstance(val, list):
        return val
    elif pd.isna(val):
        return []
    elif isinstance(val, (int, float)):
        return [int(val)]
    elif isinstance(val, str):
        try:
            return eval(val)
        except:
            return []
    return []

# Aplica a conversão para lista segura
df['Goals_Min_H'] = df['Goals_Min_H'].apply(garantir_lista)
df['Goals_Min_A'] = df['Goals_Min_A'].apply(garantir_lista)

# Função para checar se todos os gols (se existirem) foram após os 45 minutos
def nenhum_gol_primeiro_tempo(minutos):
    return all(m > 45 for m in minutos)

# Filtrar os jogos sem gols no 1º tempo
sem_gol_HT = df[
    df['Goals_Min_H'].apply(nenhum_gol_primeiro_tempo) &
    df['Goals_Min_A'].apply(nenhum_gol_primeiro_tempo)
]

# Exibir os primeiros jogos filtrados
sem_gol_HT
```

```python
# Total de jogos
total_jogos = len(df)

# Jogos sem gols no 1º tempo
jogos_sem_gol_HT = len(sem_gol_HT)

# Porcentagem
porcentagem = (jogos_sem_gol_HT / total_jogos) * 100

print(f"{porcentagem:.2f}% dos jogos terminaram 0x0 no 1º tempo.")
```

```python
# Filtra jogos com ambas as odds acima de 2
filtro_odds = sem_gol_HT[
    (sem_gol_HT['Odd_Over25_FT'] > 2) &
    (sem_gol_HT['Odd_BTTS_Yes'] > 2)
]

# Conta quantos jogos atendem aos critérios
quantidade = len(filtro_odds)

print(f"{quantidade} jogos sem gols no 1º tempo têm Odd_Over25_FT > 2 e Odd_BTTS_Yes > 2.")
```

```python
porcentagem_total = (quantidade / total_jogos) * 100

print(f"{quantidade} jogos ({porcentagem_total:.2f}%) do total terminaram 0x0 no HT e têm Odd_Over25_FT > 2 e Odd_BTTS_Yes > 2.")
```

```python
# Passo 1: Garante que Goals_Min_H e Goals_Min_A são listas seguras
def garantir_lista(val):
    if isinstance(val, list):
        return val
    elif pd.isna(val):
        return []
    elif isinstance(val, (int, float)):
        return [int(val)]
    elif isinstance(val, str):
        try:
            return eval(val)
        except:
            return []
    return []

df['Goals_Min_H'] = df['Goals_Min_H'].apply(garantir_lista)
df['Goals_Min_A'] = df['Goals_Min_A'].apply(garantir_lista)

# Passo 2: Função que verifica se não houve gols no 1º tempo
def nenhum_gol_primeiro_tempo(minutos):
    return all(m > 45 for m in minutos)

# Passo 3: Filtrar jogos com odds > 2
jogos_odds_altas = df[
    (df['Odd_Over25_FT'] > 2) &
    (df['Odd_BTTS_Yes'] > 2)
]

# Passo 4: Dentro desses jogos, verificar quais terminaram 0x0 no HT
jogos_0x0_HT_dentro_odds = jogos_odds_altas[
    jogos_odds_altas['Goals_Min_H'].apply(nenhum_gol_primeiro_tempo) &
    jogos_odds_altas['Goals_Min_A'].apply(nenhum_gol_primeiro_tempo)
]

# Passo 5: Calcular a porcentagem
total = len(jogos_odds_altas)
com_HT_0x0 = len(jogos_0x0_HT_dentro_odds)

porcentagem = (com_HT_0x0 / total) * 100

print(f"{porcentagem:.2f}% dos jogos com odds > 2 em Over 2.5 FT e BTTS Yes terminaram 0x0 no 1º tempo.")
```
