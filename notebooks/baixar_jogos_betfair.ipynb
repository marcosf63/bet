# Baixar Jogos do Dia - Betfair

Este notebook utiliza as funções do app `bet` para baixar e analisar os jogos do dia da Betfair.

```python
# Importar as bibliotecas necessárias
import sys
import os
from datetime import date, datetime
import pandas as pd
from pathlib import Path

# Adicionar o diretório raiz ao path para importar o módulo bet
project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))

# Importar funções do app bet
from bet.cli import _get_daily_games, _filter_games, _save_csv_to_file
from bet.utils import print_table
from bet.config import settings
```

## 1. Configuração e Parâmetros

```python
# Configurar data (hoje por padrão)
data_jogos = date.today().strftime("%Y-%m-%d")
fonte = "betfair"  # Pode ser: betfair, footystats, flashscore

print(f"Data dos jogos: {data_jogos}")
print(f"Fonte de dados: {fonte}")
print(f"Configurações do projeto: {settings.data_dir}")
```

## 2. Baixar Jogos do Dia

```python
# Baixar os jogos do dia usando a função do CLI
print(f"Baixando jogos do dia {data_jogos} da fonte {fonte}...")
jogos_brutos = _get_daily_games(data_jogos, fonte)

print(f"Total de jogos encontrados: {len(jogos_brutos)}")
if jogos_brutos:
    print(f"Colunas disponíveis: {list(jogos_brutos[0].keys())}")
```

## 3. Filtrar e Processar os Dados

```python
# Filtrar jogos para mostrar apenas as colunas principais
jogos_filtrados = _filter_games(jogos_brutos, fonte)

print(f"Jogos filtrados: {len(jogos_filtrados)}")
if jogos_filtrados:
    print(f"Colunas filtradas: {list(jogos_filtrados[0].keys())}")
```

## 4. Visualizar os Dados

```python
# Mostrar uma amostra dos dados em formato de tabela
if jogos_filtrados:
    print("Primeiros 10 jogos:")
    print_table(jogos_filtrados[:10])
else:
    print("Nenhum jogo encontrado para a data especificada.")
```

## 5. Converter para DataFrame do Pandas

```python
# Converter os dados para um DataFrame do pandas para análises mais avançadas
if jogos_filtrados:
    df_jogos = pd.DataFrame(jogos_filtrados)
    
    print(f"Shape do DataFrame: {df_jogos.shape}")
    print("\nPrimeiras 5 linhas:")
    display(df_jogos.head())
    
    print("\nInformações sobre o DataFrame:")
    print(df_jogos.info())
else:
    df_jogos = pd.DataFrame()
    print("DataFrame vazio - nenhum jogo encontrado.")
```

## 6. Análises Rápidas

```python
if not df_jogos.empty:
    # Contagem por liga
    print("Jogos por Liga:")
    ligas_count = df_jogos['League'].value_counts()
    print(ligas_count.head(10))
    
    # Estatísticas das odds (se disponíveis)
    if fonte == "betfair":
        odds_cols = ['Odd_H_Back', 'Odd_D_Back', 'Odd_A_Back', 'Odd_Over25_FT_Back', 'Odd_BTTS_Yes_Back']
        for col in odds_cols:
            if col in df_jogos.columns:
                df_jogos[col] = pd.to_numeric(df_jogos[col], errors='coerce')
        
        print("\nEstatísticas das Odds:")
        display(df_jogos[odds_cols].describe())
```

## 7. Filtros Personalizados

```python
if not df_jogos.empty and fonte == "betfair":
    # Exemplo: Filtrar jogos com favoritos claros (odd casa <= 1.5)
    df_jogos['Odd_H_Back_num'] = pd.to_numeric(df_jogos['Odd_H_Back'], errors='coerce')
    favoritos = df_jogos[df_jogos['Odd_H_Back_num'] <= 1.5]
    
    print(f"Jogos com favoritos claros (odd casa <= 1.5): {len(favoritos)}")
    if len(favoritos) > 0:
        display(favoritos[['League', 'Home', 'Away', 'Odd_H_Back', 'Odd_A_Back']].head(10))
    
    # Exemplo: Jogos com high over 2.5 (odd > 2.0)
    df_jogos['Odd_Over25_FT_Back_num'] = pd.to_numeric(df_jogos['Odd_Over25_FT_Back'], errors='coerce')
    high_over = df_jogos[df_jogos['Odd_Over25_FT_Back_num'] > 2.0]
    
    print(f"\nJogos com odd Over 2.5 > 2.0: {len(high_over)}")
    if len(high_over) > 0:
        display(high_over[['League', 'Home', 'Away', 'Odd_Over25_FT_Back', 'Odd_BTTS_Yes_Back']].head(10))
```

## 8. Salvar os Dados

```python
# Salvar os dados em CSV usando a função do CLI
salvar = True  # Mude para False se não quiser salvar

if salvar and jogos_filtrados:
    _save_csv_to_file(jogos_filtrados, data_jogos)
    print(f"Dados salvos na pasta data/mday/ com o nome {data_jogos}.csv")
    
    # Também salvar o DataFrame para análises futuras
    notebook_data_dir = Path.cwd() / "dados"
    notebook_data_dir.mkdir(exist_ok=True)
    
    df_file = notebook_data_dir / f"jogos_betfair_{data_jogos}.csv"
    if not df_jogos.empty:
        df_jogos.to_csv(df_file, index=False)
        print(f"DataFrame salvo em: {df_file}")
else:
    print("Salvamento desabilitado ou nenhum dado para salvar.")
```

## 9. Função Personalizada para Outras Datas

```python
def baixar_jogos_data_especifica(data_str, fonte="betfair", salvar=False):
    """
    Função para baixar jogos de uma data específica.
    
    Args:
        data_str (str): Data no formato AAAA-MM-DD
        fonte (str): Fonte dos dados (betfair, footystats, flashscore)
        salvar (bool): Se deve salvar os dados em CSV
    
    Returns:
        pd.DataFrame: DataFrame com os jogos
    """
    print(f"Baixando jogos de {data_str} da fonte {fonte}...")
    
    # Baixar dados
    jogos_brutos = _get_daily_games(data_str, fonte)
    jogos_filtrados = _filter_games(jogos_brutos, fonte)
    
    # Converter para DataFrame
    df = pd.DataFrame(jogos_filtrados) if jogos_filtrados else pd.DataFrame()
    
    # Salvar se solicitado
    if salvar and not df.empty:
        _save_csv_to_file(jogos_filtrados, data_str)
    
    print(f"Total de jogos encontrados: {len(df)}")
    return df

# Exemplo de uso:
# df_ontem = baixar_jogos_data_especifica("2024-12-27", "betfair", True)
```
