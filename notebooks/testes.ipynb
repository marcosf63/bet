```python
# Obter os IDs dos mercados
market_ids = [market_catalogue.market_id for market_catalogue in market_catalogues]
```

```python
market_names_dict = {}
market_ids_dict = {}
for market_catalogue in market_catalogues:
    market_names_dict[market_catalogue.market_id] = market_catalogue.market_name
    market_ids_dict[market_catalogue.market_name] = market_catalogue.market_id
```

```python
market_names_dict
```

```python
market_ids_selected = [
    market_ids_dict['Match Odds'],
    market_ids_dict['Both teams to Score?'],
    market_ids_dict['Correct Score'],
    market_ids_dict['Over/Under 2.5 Goals'],
    market_ids_dict['Half Time'],
    market_ids_dict['Half Time Score'],
]
```

```python
# Obter as odds atuais para os mercados
market_books = trading.betting.list_market_book(
    market_ids=market_ids
)
```

```python
for market_book in market_books:
    if market_names_dict[market_book.market_id] == 'Correct Score':
        
        print("Market:", market_names_dict[market_book.market_id])

        print("Market ID:", market_book.market_id)
        print("Status do Mercado:", market_book.status)
        print("Runners:")
        for runner in market_book.runners:
            print(" - Runner Name:", runner_names[runner.selection_id])
            if runner.last_price_traded:
                print("   - Último Preço Negociado:", runner.last_price_traded)
            else:
                print("   - Sem informação de preço")
    
        print()
```

# Buscando também as odds do lay 

```python
# Obter as odds atuais para os mercados
market_books = trading.betting.list_market_book(
    market_ids=market_ids_selected,
    price_projection={'priceData': ['EX_BEST_OFFERS']}
)

for market_book in market_books:
    print("Market ID:", market_names_dict[market_book.market_id])
    print("Status do Mercado:", market_book.status)
    print("Runners:")
    for runner in market_book.runners:
        print(" - Runner Name:", runner_names[runner.selection_id])
        # Odds de Back
        if runner.ex.available_to_back:
            back_odds = runner.ex.available_to_back[0].price  # Primeiro preço na lista de back
            print("   - Odd de Back:", back_odds)
        else:
            print("   - Sem informação de odd de back")

        # Odds de Lay
        if runner.ex.available_to_lay:
            lay_odds = runner.ex.available_to_lay[0].price  # Primeiro preço na lista de lay
            print("   - Odd de Lay:", lay_odds)
        else:
            print("   - Sem informação de odd de lay")
    print()
```
