```python
import matplotlib.pyplot as plt

# Coordenadas dos nós
nodes = {
    "16": (0, 0),
    "7.6": (-1, 1),
    "9.6": (1, 1),
    "10": (0, 2),
    "6": (-1.5, 3),
    "17": (1.5, 3),
    "8.4": (-1.5, 4),
    "25": (1.5, 4),
    "44": (2, 4),
    "160": (3, 4),
}

# Conexões entre os nós
edges = [
    ("16", "7.6"),
    ("16", "9.6"),
    ("10", "6"),
    ("10", "17"),
    ("7.6", "8.4"),
    ("9.6", "25"),
    ("17", "44"),
    ("44", "160"),
]

# Criação do gráfico
plt.figure(figsize=(8, 6))

# Desenhar arestas
for edge in edges:
    x_values = [nodes[edge[0]][0], nodes[edge[1]][0]]
    y_values = [nodes[edge[0]][1], nodes[edge[1]][1]]
    plt.plot(x_values, y_values, 'k-', alpha=0.6)

# Desenhar nós
for node, (x, y) in nodes.items():
    plt.scatter(x, y, color="orange", zorder=2)
    plt.text(x, y + 0.1, node, fontsize=9, ha="center", zorder=3)

plt.title("Árvore Exemplo", fontsize=14)
plt.axis("off")
plt.show()
```
