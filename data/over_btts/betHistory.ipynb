```python
import pandas as pd
import locale
locale.setlocale(locale.LC_TIME, 'pt_BR.UTF-8')
```

```python
file = "BetHistory16a23-12-2024.csv"
```

```python
def nomarliza_coluna_lp(value: str) -> float:
    "Coverte a coluna L/P de string para float"
    if '(' in value:
        value = value.replace('(','').replace(')', '')
        return -float(value)
    return float(value)

def prepara_dados_btts(file_path: str) -> None:
    df = pd.read_csv(file_path)
    df['Mercado'] = df['Mercado'].astype(str)
    df['Seleção'] = df['Seleção'].astype(str)
    colunas = ['Mercado', 'Seleção', 'Aposta realizada', 'Aposta ($)', 'Média de probabilidades correspondidas', 'Lucro/Prejuízo ($)']
    df_aux = df[colunas]
    df_btts = df_aux[df_aux['Seleção'] == 'Sim']
    df_btts = df_btts.rename(columns={
        'Aposta realizada': 'Data',
        'Aposta ($)': 'Stake',
        'Média de probabilidades correspondidas': 'Odd',
        'Lucro/Prejuízo ($)': 'L/P',
    })
    df_btts['L/P'] = df_btts['L/P'].apply(nomarliza_coluna_lp)
    df_btts['Data'] = pd.to_datetime(df_btts['Data'], format='%d-%b-%y %H:%M')
    df_btts['Data'] = df_btts['Data'].dt.date
    return df_btts
    
    
```

```python
btts = prepara_dados_btts(file)
```
